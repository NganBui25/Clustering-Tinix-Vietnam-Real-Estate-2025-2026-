### Điền khuyết thiếu cho frontage_width

```python
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.linear_model import BayesianRidge

print("BẮT ĐẦU QUÁ TRÌNH ĐIỀN KHUYẾT CHO FRONTAGE_WIDTH...\n")

# ==========================================
# BƯỚC 1: BÀN TAY SẮT - ÉP CHUNG CƯ VỀ 0
# ==========================================
# Áp dụng cho cả Train và Test
df_train.loc[df_train['property_type_name'] == 'Căn hộ chung cư', 'frontage_width'] = 0
df_test.loc[df_test['property_type_name'] == 'Căn hộ chung cư', 'frontage_width'] = 0

print("✅ Đã ép toàn bộ frontage_width của 'Căn hộ chung cư' về 0.")

# ==========================================
# BƯỚC 2: SỬ DỤNG MICE CHO NHÀ MẶT ĐẤT & ĐẤT
# ==========================================
# 1. Xác định danh sách các biến đưa vào MICE
mice_features = ['price', 'area', 'frontage_width']

# 2. Tạo Mask (mặt nạ) để chỉ lọc ra các BĐS không phải Chung cư
mask_train_houses = df_train['property_type_name'] != 'Căn hộ chung cư'
mask_test_houses = df_test['property_type_name'] != 'Căn hộ chung cư'

# Tách dữ liệu để đưa vào mô hình
train_data_for_mice = df_train.loc[mask_train_houses, mice_features]
test_data_for_mice = df_test.loc[mask_test_houses, mice_features]

# 3. Khởi tạo mô hình MICE với BayesianRidge để tối ưu tốc độ
mice_imputer = IterativeImputer(
    estimator=BayesianRidge(), 
    max_iter=10, 
    random_state=42
)

# 4. Huấn luyện (Fit) và Điền khuyết (Transform) trên tập TRAIN
print("⏳ Đang huấn luyện MICE trên tập Train... (Có thể mất vài phút)")
train_imputed_array = mice_imputer.fit_transform(train_data_for_mice)
df_train.loc[mask_train_houses, mice_features] = train_imputed_array

# 5. Điền khuyết (Transform) trên tập TEST (Tuyệt đối KHÔNG fit lại)
print("⏳ Đang áp dụng MICE sang tập Test...")
test_imputed_array = mice_imputer.transform(test_data_for_mice)
df_test.loc[mask_test_houses, mice_features] = test_imputed_array

print("✅ Đã điền khuyết frontage_width bằng MICE thành công!\n")

# ==========================================
# BƯỚC 3: KIỂM TRA LẠI KẾT QUẢ
# ==========================================
print("--- KẾT QUẢ KIỂM TRA ---")
print(f"Số lượng null mặt tiền trên Train: {df_train['frontage_width'].isnull().sum()}")
print(f"Số lượng null mặt tiền trên Test: {df_test['frontage_width'].isnull().sum()}")
```

### Điền khuyết thiếu cho count_floor

```python
import pandas as pd
import numpy as np

print("BẮT ĐẦU ĐIỀN KHUYẾT CHO FLOOR_COUNT...\n")

# Danh sách nhóm gán = 0
zero_floor_groups = ['Đất', 'Căn hộ chung cư']

# ==========================================
# BƯỚC 1: BÀN TAY SẮT - ÉP ĐẤT & CHUNG CƯ VỀ 0
# ==========================================
df_train.loc[df_train['property_type_name'].isin(zero_floor_groups), 'floor_count'] = 0
df_test.loc[df_test['property_type_name'].isin(zero_floor_groups), 'floor_count'] = 0

print("✅ Đã ép số tầng của Đất và Chung cư về 0.")

# ==========================================
# BƯỚC 2: GROUP-BASED MEDIAN CHO NHÀ, BIỆT THỰ, SHOPHOUSE
# ==========================================
# Lọc ra nhóm cần tính Median (bỏ Đất và Chung cư)
mask_train_calc = ~df_train['property_type_name'].isin(zero_floor_groups)

# Học (Fit) mapping từ tập Train theo Loại BĐS và Quận
floor_medians = df_train[mask_train_calc].groupby(
    ['property_type_name', 'district_name']
)['floor_count'].median().reset_index()

floor_medians.rename(columns={'floor_count': 'floor_median'}, inplace=True)

# --- Transform trên Train ---
df_train = df_train.merge(floor_medians, on=['property_type_name', 'district_name'], how='left')
df_train['floor_count'] = df_train['floor_count'].fillna(df_train['floor_median'])
df_train = df_train.drop(columns=['floor_median'])

# --- Transform trên Test ---
df_test = df_test.merge(floor_medians, on=['property_type_name', 'district_name'], how='left')
df_test['floor_count'] = df_test['floor_count'].fillna(df_test['floor_median'])
df_test = df_test.drop(columns=['floor_median'])

# ==========================================
# BƯỚC 3: FALLBACK & LÀM TRÒN THÀNH SỐ NGUYÊN
# ==========================================
# Fallback: Dùng Median của toàn bộ Loại hình (nếu Quận đó thiếu dữ liệu)
global_floor = df_train.groupby('property_type_name')['floor_count'].median()

df_train['floor_count'] = df_train['floor_count'].fillna(df_train['property_type_name'].map(global_floor))
df_test['floor_count'] = df_test['floor_count'].fillna(df_test['property_type_name'].map(global_floor))

# Làm tròn thành số nguyên và ép kiểu Int
df_train['floor_count'] = np.round(df_train['floor_count']).astype(int)
df_test['floor_count'] = np.round(df_test['floor_count']).astype(int)

print("✅ Đã hoàn thành điền khuyết theo Quận/Huyện và làm tròn số tầng.\n")

# KIỂM TRA LẠI
print("--- KẾT QUẢ KIỂM TRA ---")
print(f"Train - Null Số tầng: {df_train['floor_count'].isnull().sum()}")
print(f"Test  - Null Số tầng: {df_test['floor_count'].isnull().sum()}")
```

### Điền khuyết thiếu cho count_bedroom và count_bathroom

```python
import pandas as pd
import numpy as np

print("BẮT ĐẦU ĐIỀN KHUYẾT CHO PHÒNG NGỦ & PHÒNG TẮM...\n")

# ==========================================
# BƯỚC 1: ĐIỀN KHUYẾT CHÉO TỪ BIẾN "ANH EM" TƯƠNG QUAN 0.93
# ==========================================
# Học (Fit) mapping từ tập Train
bed_to_bath_map = df_train.groupby('bedroom_count')['bathroom_count'].median()
bath_to_bed_map = df_train.groupby('bathroom_count')['bedroom_count'].median()

# Điền (Transform) cho tập Train
df_train['bathroom_count'] = df_train['bathroom_count'].fillna(df_train['bedroom_count'].map(bed_to_bath_map))
df_train['bedroom_count'] = df_train['bedroom_count'].fillna(df_train['bathroom_count'].map(bath_to_bed_map))

# Điền (Transform) cho tập Test
df_test['bathroom_count'] = df_test['bathroom_count'].fillna(df_test['bedroom_count'].map(bed_to_bath_map))
df_test['bedroom_count'] = df_test['bedroom_count'].fillna(df_test['bathroom_count'].map(bath_to_bed_map))

print("✅ Đã hoàn thành điền khuyết chéo giữa 2 biến.")

# ==========================================
# BƯỚC 2: GROUP-BASED MEDIAN (XỬ LÝ DÒNG BỊ KHUYẾT CẢ HAI)
# ==========================================
# Học (Fit) mapping từ tập Train theo Loại BĐS và Quận
room_medians = df_train.groupby(['property_type_name', 'district_name'])[['bedroom_count', 'bathroom_count']].median().reset_index()
room_medians.rename(columns={'bedroom_count': 'bed_median', 'bathroom_count': 'bath_median'}, inplace=True)

# --- Transform trên Train ---
df_train = df_train.merge(room_medians, on=['property_type_name', 'district_name'], how='left')
df_train['bedroom_count'] = df_train['bedroom_count'].fillna(df_train['bed_median'])
df_train['bathroom_count'] = df_train['bathroom_count'].fillna(df_train['bath_median'])
df_train = df_train.drop(columns=['bed_median', 'bath_median'])

# --- Transform trên Test ---
df_test = df_test.merge(room_medians, on=['property_type_name', 'district_name'], how='left')
df_test['bedroom_count'] = df_test['bedroom_count'].fillna(df_test['bed_median'])
df_test['bathroom_count'] = df_test['bathroom_count'].fillna(df_test['bath_median'])
df_test = df_test.drop(columns=['bed_median', 'bath_median'])

# ==========================================
# BƯỚC 3: FALLBACK & LÀM TRÒN THÀNH SỐ NGUYÊN
# ==========================================
# Fallback cuối cùng nếu vẫn còn null (do Quận đó không có dữ liệu)
global_bed = df_train['bedroom_count'].median()
global_bath = df_train['bathroom_count'].median()

for df_target in [df_train, df_test]:
    df_target['bedroom_count'] = df_target['bedroom_count'].fillna(global_bed)
    df_target['bathroom_count'] = df_target['bathroom_count'].fillna(global_bath)
    
    # Làm tròn thành số nguyên (Round) và ép kiểu Int
    df_target['bedroom_count'] = np.round(df_target['bedroom_count']).astype(int)
    df_target['bathroom_count'] = np.round(df_target['bathroom_count']).astype(int)

print("✅ Đã hoàn thành điền khuyết theo nhóm và làm tròn số nguyên.\n")

# KIỂM TRA LẠI
print("--- KẾT QUẢ KIỂM TRA ---")
print(f"Train - Null Phòng ngủ: {df_train['bedroom_count'].isnull().sum()}, Null Phòng tắm: {df_train['bathroom_count'].isnull().sum()}")
print(f"Test  - Null Phòng ngủ: {df_test['bedroom_count'].isnull().sum()}, Null Phòng tắm: {df_test['bathroom_count'].isnull().sum()}")
```

###  - Hàm vẽ boxplot trước/sau và hàm thống kê outlier

Do `price`, `area`, `frontage_width` lệch phải rất mạnh, boxplot nên dùng thang log để quan sát rõ phần lớn dữ liệu. Các biến đếm vẫn dùng thang thường.

```python
def plot_box_before_after(before_df, after_df, col, title=None, scale='linear', divide_by=1):
    """Vẽ boxplot trước/sau cho một biến."""
    before = before_df[col].dropna() / divide_by
    after = after_df[col].dropna() / divide_by

    plot_df = pd.DataFrame({
        'Giá trị': pd.concat([before, after], ignore_index=True),
        'Giai đoạn': ['Trước xử lý'] * len(before) + ['Sau xử lý'] * len(after)
    })

    plt.figure(figsize=(12, 3.8))
    sns.boxplot(data=plot_df, x='Giá trị', y='Giai đoạn', orient='h', showfliers=True)
    plt.title(title or f'Boxplot trước/sau xử lý outlier: {col}', fontsize=12, fontweight='bold')
    plt.xlabel(col)
    if scale == 'log':
        # Dùng symlog để vẫn hiển thị được giá trị 0, ví dụ chung cư có frontage_width = 0
        plt.xscale('symlog', linthresh=1)
    plt.tight_layout()
    plt.show()


def summarize_change(before_df, after_df, cols):
    """Thống kê số dòng thay đổi sau xử lý outlier."""
    rows = []
    for col in cols:
        before = before_df[col]
        after = after_df[col]
        changed = before.ne(after) & ~(before.isna() & after.isna())
        rows.append({
            'Biến': col,
            'Số dòng bị điều chỉnh': int(changed.sum()),
            'Tỷ lệ bị điều chỉnh (%)': round(changed.mean() * 100, 3),
            'Min trước': before.min(),
            'Min sau': after.min(),
            'P50 trước': before.median(),
            'P50 sau': after.median(),
            'P99 trước': before.quantile(0.99),
            'P99 sau': after.quantile(0.99),
            'Max trước': before.max(),
            'Max sau': after.max(),
        })
    return pd.DataFrame(rows)

print("Đã tạo hàm vẽ boxplot và hàm đánh giá thay đổi.")
```

### Hàm tính ngưỡng theo nhóm và clip an toàn

Ở dữ liệu BĐS, cùng một giá trị có thể là bình thường với biệt thự nhưng bất thường với căn hộ. Vì vậy ngưỡng được học theo `property_type_name`; nếu nhóm quá nhỏ thì fallback về ngưỡng toàn train.


```python
def fit_group_bounds(train_df, col, group_col='property_type_name', q_low=0.005, q_high=0.995, min_count=100):
    """Học ngưỡng winsorize từ train theo nhóm loại BĐS."""
    global_low = train_df[col].quantile(q_low)
    global_high = train_df[col].quantile(q_high)

    bounds = train_df.groupby(group_col)[col].agg(
        n='count',
        low=lambda s: s.quantile(q_low),
        high=lambda s: s.quantile(q_high)
    )

    # Nhóm quá nhỏ hoặc thiếu ngưỡng thì dùng ngưỡng global
    bounds.loc[bounds['n'] < min_count, ['low', 'high']] = [global_low, global_high]
    bounds['low'] = bounds['low'].fillna(global_low)
    bounds['high'] = bounds['high'].fillna(global_high)

    return bounds[['low', 'high']], global_low, global_high


def apply_group_clip(df_target, col, bounds, global_low, global_high, group_col='property_type_name', hard_low=None, hard_high=None):
    """Clip một cột theo ngưỡng nhóm đã học từ train."""
    lower = df_target[group_col].map(bounds['low']).fillna(global_low)
    upper = df_target[group_col].map(bounds['high']).fillna(global_high)

    if hard_low is not None:
        lower = np.maximum(lower, hard_low)
    if hard_high is not None:
        upper = np.minimum(upper, hard_high)

    df_target[col] = df_target[col].clip(lower=lower, upper=upper)
    return df_target

print("Đã tạo hàm fit/apply ngưỡng outlier theo nhóm.")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# =========================================================
# PHÂN TÍCH SENSITIVITY: Tỉ lệ bị clipping theo k
# =========================================================

def compute_clip_rate_by_k(df, col, group_col, k_values, min_count=100):
    """
    Tính tỉ lệ dữ liệu bị clipping với từng hệ số k khác nhau.
    Giúp chọn k hợp lý mà không loại nhầm dữ liệu hợp lệ.
    """
    rows = []
    for group, gdf in df.groupby(group_col):
        values = gdf[col].dropna()
        values = values[values > 0]
        if len(values) < min_count:
            continue
        log_vals = np.log1p(values)
        q1, q3 = log_vals.quantile(0.25), log_vals.quantile(0.75)
        iqr = q3 - q1
        for k in k_values:
            low = np.expm1(q1 - k * iqr)
            high = np.expm1(q3 + k * iqr)
            n_clipped = ((values < low) | (values > high)).sum()
            rows.append({
                'Nhóm BĐS': group,
                'k': k,
                'n': len(values),
                'Tỉ lệ bị clip (%)': round(n_clipped / len(values) * 100, 2),
                'Lower bound': round(low, 0),
                'Upper bound': round(high, 0)
            })
    return pd.DataFrame(rows)

k_values = [1.5, 2.0, 3.0]

print("=== SENSITIVITY ANALYSIS: PRICE ===")
price_sensitivity = compute_clip_rate_by_k(
    df_train_before_outlier, 'price', 'property_type_name', k_values
)
display(price_sensitivity.pivot_table(
    index='Nhóm BĐS', columns='k', values='Tỉ lệ bị clip (%)'
).round(2))

print("\n=== SENSITIVITY ANALYSIS: AREA ===")
area_sensitivity = compute_clip_rate_by_k(
    df_train_before_outlier, 'area', 'property_type_name', k_values
)
display(area_sensitivity.pivot_table(
    index='Nhóm BĐS', columns='k', values='Tỉ lệ bị clip (%)'
).round(2))

# =========================================================
# VISUALIZATION: clip rate theo k
# =========================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

for ax, sensitivity_df, title in zip(
    axes,
    [price_sensitivity, area_sensitivity],
    ['PRICE — Tỉ lệ clipping theo k', 'AREA — Tỉ lệ clipping theo k']
):
    for k in k_values:
        sub = sensitivity_df[sensitivity_df['k'] == k]
        ax.bar(
            [x + k_values.index(k) * 0.25 for x in range(len(sub))],
            sub['Tỉ lệ bị clip (%)'],
            width=0.22,
            label=f'k={k}'
        )
    ax.set_xticks(range(len(sub)))
    ax.set_xticklabels(sub['Nhóm BĐS'], rotation=30, ha='right')
    ax.set_ylabel('Tỉ lệ bị clipping (%)')
    ax.set_title(title, fontweight='bold')
    ax.axhline(5, color='red', linestyle='--', linewidth=1, label='Ngưỡng 5% cảnh báo')
    ax.legend()

plt.suptitle('Sensitivity Analysis: Lựa chọn hệ số k trong Log-IQR', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()


```

Dựa trên sensitivity analysis của tỷ lệ clipping theo từng hệ số k, nhóm lựa chọn k=2.0 cho phương pháp Log-IQR. Giá trị này giúp cân bằng giữa việc loại bỏ các extreme outlier và bảo toàn các phân khúc bất động sản hợp lệ có giá trị lớn, đồng thời giữ tỷ lệ clipping của đa số nhóm dưới ngưỡng cảnh báo 5%.

Ngưỡng outlier của biến price được tính riêng cho từng property_type_name bằng phương pháp Log-IQR nhằm phản ánh đúng đặc trưng giá của từng phân khúc bất động sản. Sau khi áp dụng log1p(price) để giảm độ lệch phân phối, ngưỡng dưới và ngưỡng trên được xác định theo công thức IQR với hệ số k=2 Các giá trị vượt ngoài khoảng này sẽ được clipping thay vì xóa bỏ nhằm hạn chế mất dữ liệu và giữ ổn định cho bài toán phân cụm.

```python
# PRICE: tiền tệ, lệch phải mạnh -> Log-IQR theo từng loại BĐS
# Lý do thay cho winsorize + hard_low:
# - EDA cho thấy price lệch phải rất mạnh, nên tính IQR trên log1p(price)
# - Mỗi property_type_name có phân phối giá khác nhau, nên ngưỡng phải học theo từng nhóm
# - Không hard-code 50 triệu; toàn bộ ngưỡng được học từ df_train để tránh cảm tính
# - Dùng clipping thay vì xóa dòng để không làm mất dữ liệu cho bài toán clustering

import numpy as np
import pandas as pd


def fit_group_log_iqr_bounds(df, col, group_col, k=2, min_count=100):
    """
    Học ngưỡng outlier cho biến liên tục bị lệch nặng bằng Log-IQR theo từng group.
    Ngưỡng được fit trên train, sau đó áp dụng cho train/test.
    """
    rows = []

    # Global fallback cho nhóm ít dữ liệu hoặc nhóm mới xuất hiện ở test
    valid_global = df.loc[df[col].notna() & (df[col] > 0), col]
    log_global = np.log1p(valid_global)

    g_q1 = log_global.quantile(0.25)
    g_q3 = log_global.quantile(0.75)
    g_iqr = g_q3 - g_q1

    global_low = max(0, np.expm1(g_q1 - k * g_iqr))
    global_high = np.expm1(g_q3 + k * g_iqr)

    for group, group_data in df.groupby(group_col):
        values = group_data.loc[group_data[col].notna() & (group_data[col] > 0), col]

        # Nếu nhóm quá ít dữ liệu, dùng ngưỡng global để tránh ngưỡng thiếu ổn định
        if len(values) < min_count:
            rows.append({
                group_col: group,
                'n': len(values),
                'method': 'global_log_iqr_fallback',
                'lower_bound': global_low,
                'upper_bound': global_high
            })
            continue

        log_values = np.log1p(values)
        q1 = log_values.quantile(0.25)
        q3 = log_values.quantile(0.75)
        iqr = q3 - q1

        lower = max(0, np.expm1(q1 - k * iqr))
        upper = np.expm1(q3 + k * iqr)

        rows.append({
            group_col: group,
            'n': len(values),
            'method': 'group_log_iqr',
            'lower_bound': lower,
            'upper_bound': upper
        })

    bounds = pd.DataFrame(rows)
    return bounds, global_low, global_high


def apply_group_log_iqr_clip(df, col, group_col, bounds, global_low, global_high):
    """
    Áp dụng clipping theo ngưỡng Log-IQR đã học từ train.
    Không fit lại trên test để tránh data leakage.
    """
    df = df.copy()

    bound_map = bounds.set_index(group_col)[['lower_bound', 'upper_bound']].to_dict('index')

    for group in df[group_col].dropna().unique():
        if group in bound_map:
            low = bound_map[group]['lower_bound']
            high = bound_map[group]['upper_bound']
        else:
            low = global_low
            high = global_high

        mask = df[group_col] == group
        df.loc[mask, col] = df.loc[mask, col].clip(lower=low, upper=high)

    # Các dòng thiếu group_col cũng dùng ngưỡng global
    mask_missing_group = df[group_col].isna()
    df.loc[mask_missing_group, col] = df.loc[mask_missing_group, col].clip(
        lower=global_low,
        upper=global_high
    )

    return df


# Fit ngưỡng trên train theo từng property_type_name
price_bounds, price_global_low, price_global_high = fit_group_log_iqr_bounds(
    df_train,
    col='price',
    group_col='property_type_name',
    k=1.5,
    min_count=100
)

# Apply cùng ngưỡng từ train cho train và test
df_train = apply_group_log_iqr_clip(
    df_train,
    col='price',
    group_col='property_type_name',
    bounds=price_bounds,
    global_low=price_global_low,
    global_high=price_global_high
)

df_test = apply_group_log_iqr_clip(
    df_test,
    col='price',
    group_col='property_type_name',
    bounds=price_bounds,
    global_low=price_global_low,
    global_high=price_global_high
)

print("Ngưỡng PRICE học từ train bằng Log-IQR theo property_type_name:")
display(price_bounds)

print("Global fallback lower:", format(int(price_global_low), ','))
print("Global fallback upper:", format(int(price_global_high), ','))

# =========================================================
# VISUALIZATION ĐÚNG VỚI KỸ THUẬT GROUP-WISE
# Vì price được xử lý theo property_type_name, đánh giá cũng phải vẽ theo từng nhóm.
# Dùng log1p(price) để nhìn rõ phân phối lệch phải.
# =========================================================

import matplotlib.pyplot as plt
import seaborn as sns


def plot_group_box_before_after(before_df, after_df, col, group_col, sample_n=60000):
    """
    Vẽ boxenplot trước/sau theo từng nhóm để đánh giá xử lý outlier group-wise.
    Dùng sample để tránh notebook bị chậm khi dữ liệu lớn.
    """
    before_sample = before_df.sample(min(sample_n, len(before_df)), random_state=42).copy()
    after_sample = after_df.sample(min(sample_n, len(after_df)), random_state=42).copy()

    before_plot = before_sample[[group_col, col]].dropna().copy()
    after_plot = after_sample[[group_col, col]].dropna().copy()

    before_plot = before_plot[before_plot[col] > 0]
    after_plot = after_plot[after_plot[col] > 0]

    before_plot['Giai đoạn'] = 'Trước xử lý'
    after_plot['Giai đoạn'] = 'Sau xử lý'

    plot_df = pd.concat([before_plot, after_plot], ignore_index=True)
    plot_df[group_col] = plot_df[group_col].fillna('Không rõ')
    plot_df[f'log1p_{col}'] = np.log1p(plot_df[col])

    order = (
        before_df[group_col]
        .fillna('Không rõ')
        .value_counts()
        .index
        .tolist()
    )

    plt.figure(figsize=(15, 6))

    sns.boxenplot(
        data=plot_df,
        x=group_col,
        y=f'log1p_{col}',
        hue='Giai đoạn',
        order=order
    )

    plt.title(
        'Price trước/sau xử lý outlier theo từng property_type_name - Log-IQR',
        fontsize=12,
        fontweight='bold'
    )
    plt.xlabel('Loại hình bất động sản')
    plt.ylabel('log1p(price)')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()


plot_group_box_before_after(
    df_train_before_outlier,
    df_train,
    col='price',
    group_col='property_type_name',
    sample_n=60000
)

print("""
Đánh giá: Vì ngưỡng Log-IQR của price được học riêng theo từng property_type_name,
boxplot/boxenplot sau xử lý cũng được trực quan hóa theo từng nhóm. Cách này giúp kiểm tra
rõ nhóm nào bị clipping nhiều, nhóm nào giữ nguyên tail hợp lý, và tránh hiểu nhầm rằng
price đang được xử lý bằng một ngưỡng global cho toàn bộ dữ liệu.
""")

```

###  Xử lý outlier cho `area`

`area` phải dương và rất lệch phải. Một số dòng diện tích cực lớn thường là lỗi nhập hoặc tin dự án/đất rất lớn làm méo phân cụm. Dùng luật sàn thực tế theo loại hình, sau đó winsorize đỉnh theo nhóm loại BĐS.


Biến area được xử lý outlier bằng kỹ thuật Log-IQR theo từng property_type_name nhằm phản ánh đúng sự khác biệt diện tích giữa các phân khúc bất động sản. Phép biến đổi log1p() giúp giảm độ lệch phải của phân phối, sau đó IQR được tính riêng cho từng nhóm để tránh loại nhầm các bất động sản có diện tích lớn nhưng hợp lệ như biệt thự hoặc đất nền. Các giá trị vượt ngưỡng được clipping thay vì xóa bỏ nhằm bảo toàn dữ liệu cho bài toán phân cụm.

```python
# =========================================================
# AREA: xử lý outlier bằng Log-IQR theo property_type_name
# =========================================================

# Lý do:
# - Area lệch phải mạnh (heavy-tail)
# - Mỗi loại BĐS có phân phối diện tích khác nhau
# - Không dùng global IQR để tránh xóa nhầm biệt thự/đất lớn
# - Dùng clipping thay vì remove để không mất dữ liệu clustering

import numpy as np
import pandas as pd


# backup dữ liệu trước xử lý
df_train_before_area_outlier = df_train.copy()


# =========================================================
# HÀM FIT NGƯỠNG LOG-IQR THEO GROUP
# =========================================================

def fit_group_log_iqr_bounds_area(
    df,
    col,
    group_col,
    k=2,
    min_count=100,
    min_floor=8.0
):
    """
    Học ngưỡng Log-IQR theo từng property_type_name.
    """

    rows = []

    # =====================================================
    # GLOBAL FALLBACK
    # =====================================================

    valid_global = df.loc[
        df[col].notna() & (df[col] > 0),
        col
    ]

    log_global = np.log1p(valid_global)

    g_q1 = log_global.quantile(0.25)
    g_q3 = log_global.quantile(0.75)

    g_iqr = g_q3 - g_q1

    global_low = max(
        min_floor,
        np.expm1(g_q1 - k * g_iqr)
    )

    global_high = np.expm1(
        g_q3 + k * g_iqr
    )

    # =====================================================
    # GROUP-WISE IQR
    # =====================================================

    for group, group_data in df.groupby(group_col):

        values = group_data.loc[
            group_data[col].notna() & (group_data[col] > 0),
            col
        ]

        # nhóm ít dữ liệu -> fallback global
        if len(values) < min_count:

            rows.append({
                group_col: group,
                'n': len(values),
                'method': 'global_log_iqr_fallback',
                'lower_bound': global_low,
                'upper_bound': global_high
            })

            continue

        # log transform
        log_values = np.log1p(values)

        q1 = log_values.quantile(0.25)
        q3 = log_values.quantile(0.75)

        iqr = q3 - q1

        lower = max(
            min_floor,
            np.expm1(q1 - k * iqr)
        )

        upper = np.expm1(
            q3 + k * iqr
        )

        rows.append({
            group_col: group,
            'n': len(values),
            'method': 'group_log_iqr',
            'lower_bound': lower,
            'upper_bound': upper
        })

    bounds = pd.DataFrame(rows)

    return bounds, global_low, global_high


# =========================================================
# APPLY CLIPPING
# =========================================================

def apply_group_log_iqr_clip_area(
    df,
    col,
    group_col,
    bounds,
    global_low,
    global_high
):

    df = df.copy()

    bound_map = bounds.set_index(group_col)[
        ['lower_bound', 'upper_bound']
    ].to_dict('index')

    for group in df[group_col].dropna().unique():

        if group in bound_map:

            low = bound_map[group]['lower_bound']
            high = bound_map[group]['upper_bound']

        else:

            low = global_low
            high = global_high

        mask = df[group_col] == group

        df.loc[mask, col] = df.loc[
            mask,
            col
        ].clip(lower=low, upper=high)

    # group missing
    mask_missing_group = df[group_col].isna()

    df.loc[mask_missing_group, col] = df.loc[
        mask_missing_group,
        col
    ].clip(
        lower=global_low,
        upper=global_high
    )

    return df


# =========================================================
# FIT NGƯỠNG TỪ TRAIN
# =========================================================

area_bounds, area_global_low, area_global_high = fit_group_log_iqr_bounds_area(
    df_train,
    col='area',
    group_col='property_type_name',
    k=1.5,
    min_count=100,
    min_floor=8.0
)


# =========================================================
# APPLY CHO TRAIN/TEST
# =========================================================

df_train = apply_group_log_iqr_clip_area(
    df_train,
    col='area',
    group_col='property_type_name',
    bounds=area_bounds,
    global_low=area_global_low,
    global_high=area_global_high
)

df_test = apply_group_log_iqr_clip_area(
    df_test,
    col='area',
    group_col='property_type_name',
    bounds=area_bounds,
    global_low=area_global_low,
    global_high=area_global_high
)


# =========================================================
# HIỂN THỊ NGƯỠNG
# =========================================================

print("Ngưỡng AREA học từ train bằng Log-IQR theo property_type_name:")

display(area_bounds)

print("Global fallback lower:", round(area_global_low, 2))
print("Global fallback upper:", round(area_global_high, 2))


# =========================================================
# VISUALIZATION THEO PROPERTY TYPE
# =========================================================

import matplotlib.pyplot as plt
import seaborn as sns

sample_before = df_train_before_area_outlier.sample(
    min(50000, len(df_train_before_area_outlier)),
    random_state=42
)

sample_after = df_train.sample(
    min(50000, len(df_train)),
    random_state=42
)

sample_before['area_log'] = np.log1p(sample_before['area'])
sample_after['area_log'] = np.log1p(sample_after['area'])

fig, axes = plt.subplots(2, 2, figsize=(20, 12))


# BEFORE AREA
sns.boxenplot(
    data=sample_before,
    x='area',
    y='property_type_name',
    ax=axes[0,0],
    palette='Set3'
)

axes[0,0].set_title('Area gốc trước xử lý')
axes[0,0].set_xlabel('Area (m²)')


# AFTER AREA
sns.boxenplot(
    data=sample_after,
    x='area',
    y='property_type_name',
    ax=axes[0,1],
    palette='Set3'
)

axes[0,1].set_title('Area gốc sau xử lý')
axes[0,1].set_xlabel('Area (m²)')


# BEFORE LOG AREA
sns.boxenplot(
    data=sample_before,
    x='area_log',
    y='property_type_name',
    ax=axes[1,0],
    palette='Set3'
)

axes[1,0].set_title('log(area) trước xử lý')
axes[1,0].set_xlabel('log1p(area)')


# AFTER LOG AREA
sns.boxenplot(
    data=sample_after,
    x='area_log',
    y='property_type_name',
    ax=axes[1,1],
    palette='Set3'
)

axes[1,1].set_title('log(area) sau xử lý')
axes[1,1].set_xlabel('log1p(area)')


plt.suptitle(
    'So sánh phân phối Area trước/sau xử lý theo từng loại BĐS',
    fontsize=16
)

plt.tight_layout()

plt.show()


# cleanup
sample_before.drop(columns=['area_log'], inplace=True, errors='ignore')
sample_after.drop(columns=['area_log'], inplace=True, errors='ignore')
```

Sau khi xử lý outlier cho area bằng phương pháp phân nhóm kết hợp biến đổi Log, dữ liệu đã sạch hoàn toàn các giá trị dị biệt và đưa các đường râu về biên an toàn. Đồng thời, biểu đồ Log giúp loại bỏ hiện tượng nén dữ liệu của thang đo gốc, nhìn rõ dải diện tích tăng tiến cực kỳ logic theo từng loại hình bất động sản.

###  Xử lý outlier cho `frontage_width`

`frontage_width` có ý nghĩa với nhà/đất/biệt thự/shophouse, nhưng với căn hộ bạn đã quy ước bằng 0. 

Sử dụng kỹ thuật phân nhóm kết hợp chặn biên (Grouped Clipping) dựa trên phân vị toán học từ $0.5\%$ đến $99.5\%$ của từng loại hình bất động sản để loại bỏ triệt để lỗi nhập liệu cho biến frontage_width. Cụ thể, thuật toán cô lập và gán cứng nhóm chung cư về bằng 0 vì loại hình này không có mặt tiền riêng biệt. Với các nhóm nhà đất còn lại, dữ liệu được ép biên dưới theo giới hạn vật lý thực tế (như tối thiểu $1.5m$ cho nhà/đất và $3.0m$ cho biệt thự/shophouse) và ép biên trên theo ngưỡng trần của từng phân khúc, giúp làm sạch nhiễu mà vẫn giữ vẹn toàn cấu trúc dòng dữ liệu cho cả tập train và test.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =====================================================================
# 1. HỌC NGƯỠNG BIÊN TỪ TẬP TRAIN (LOẠI TRỪ CHUNG CƯ)
# =====================================================================
frontage_bounds, frontage_global_low, frontage_global_high = fit_group_bounds(
    df_train[df_train['property_type_name'] != 'Căn hộ chung cư'],
    col='frontage_width', group_col='property_type_name', q_low=0.005, q_high=0.995, min_count=100
)

# Cấu hình ngưỡng chặn dưới vật lý thực tế
frontage_min_by_type = {
    'Nhà': 1.5,
    'Đất': 1.5,
    'Biệt thự/Nhà liền kề': 3.0,
    'Shophouse': 3.0
}

# =====================================================================
# 2. HÀM XỬ LÝ OUTLIER THEO NHÓM VÀ LÀM TRÒN SỐ
# =====================================================================
def apply_frontage_clip(df_target):
    df_res = df_target.copy()
    
    # Ép nhóm Căn hộ chung cư về bằng 0 (vì chung cư không có mặt tiền riêng)
    condo_mask = df_res['property_type_name'] == 'Căn hộ chung cư'
    df_res.loc[condo_mask, 'frontage_width'] = 0

    # Xử lý chặn biên (Clipping) cho các nhóm nhà đất còn lại
    non_condo = ~condo_mask
    lower = df_res.loc[non_condo, 'property_type_name'].map(frontage_bounds['low']).fillna(frontage_global_low)
    upper = df_res.loc[non_condo, 'property_type_name'].map(frontage_bounds['high']).fillna(frontage_global_high)
    
    # Áp dụng luật chặn dưới vật lý
    hard_lower = df_res.loc[non_condo, 'property_type_name'].map(frontage_min_by_type).fillna(1.5)
    lower = np.maximum(lower, hard_lower)

    # Ép biên và làm tròn 1 chữ số thập phân cho dữ liệu đồng bộ
    df_res.loc[non_condo, 'frontage_width'] = df_res.loc[non_condo, 'frontage_width'].clip(lower=lower, upper=upper).round(1)
    return df_res

# Thực thi làm sạch dữ liệu
df_train = apply_frontage_clip(df_train)
df_test = apply_frontage_clip(df_test)

# =====================================================================
# 3. CODE TRỰC QUAN HÓA PHÂN NHÓM (SỬA LỖI TRỤC ÂM)
# =====================================================================
# Tạo bản sao tính toán log1p trực tiếp để đồ thị hiển thị chuẩn xác
df_plot_before = df_train_before_outlier.copy()
df_plot_before['log_frontage'] = np.log1p(df_plot_before['frontage_width'])

df_plot_after = df_train.copy()
df_plot_after['log_frontage'] = np.log1p(df_plot_after['frontage_width'])

# Cấu hình lưới hiển thị song song 2 biểu đồ
fig, axes = plt.subplots(1, 2, figsize=(16, 6), sharey=True)

# Biểu đồ 1: Trước xử lý phân nhóm
sns.boxplot(data=df_plot_before, x='log_frontage', y='property_type_name', ax=axes[0], palette='Set2')
axes[0].set_title('Frontage Width Trước Xử Lý (Phân Nhóm)', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Thang đo log1p(frontage_width)', fontsize=11)
axes[0].set_ylabel('Loại hình bất động sản', fontsize=11)

# Biểu đồ 2: Sau khi gọt sạch Outlier theo nhóm
sns.boxplot(data=df_plot_after, x='log_frontage', y='property_type_name', ax=axes[1], palette='Set2')
axes[1].set_title('Frontage Width Sau Xử Lý (Phân Nhóm)', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Thang đo log1p(frontage_width)', fontsize=11)
axes[1].set_ylabel('')

plt.tight_layout()
plt.show()
```

Biểu đồ này cho thấy kỹ thuật xử lý frontage_width theo từng property_type_name đang hoạt động khá tốt. Trước xử lý, phân phối của nhiều nhóm xuất hiện rất nhiều outlier kéo dài về phía phải, đặc biệt ở nhóm Đất và Nhà, cho thấy dữ liệu bị lệch mạnh và tồn tại nhiều giá trị bất thường. Sau khi áp dụng Log-IQR theo nhóm, các boxplot trở nên gọn và ổn định hơn, phần lớn extreme values đã được ép về ngưỡng hợp lý trong khi vẫn giữ được sự khác biệt tự nhiên giữa các loại hình bất động sản.

###  Xử lý outlier cho `floor_count`

```python
# BEDROOM_COUNT & BATHROOM_COUNT: biến đếm -> Đất = 0, loại khác clip theo nhóm + luật thực tế
room_zero_groups = ['Đất']
room_cols = ['bedroom_count', 'bathroom_count']

room_hard_high = {
    'bedroom_count': {
        'Căn hộ chung cư': 4,          # Rất hiếm căn hộ > 4 phòng ngủ (trừ penthouse)
        'Nhà': 8,                      # Nhà phố 4-5 tầng, tối đa khoảng 8 phòng là hợp lý
        'Biệt thự/Nhà liền kề': 8,     # Biệt thự quy chuẩn thường từ 4-6 phòng ngủ, chặn 8 là vừa đẹp
        'Shophouse': 6                 # Shophouse chủ yếu kinh doanh, phần ở chỉ khoảng 3-6 phòng
    },
    'bathroom_count': {
        'Căn hộ chung cư': 4,          # Đồng bộ với số phòng ngủ lớn nhất của chung cư
        'Nhà': 8,                      # Mỗi tầng 1-2 phòng vệ sinh
        'Biệt thự/Nhà liền kề': 8,     # Số phòng tắm biệt thự thường tương đương hoặc hơn số phòng ngủ 1 phòng
        'Shophouse': 6                 # Thiết kế vệ sinh theo tầng
    }
}

room_bounds_dict = {}
for col in room_cols:
    bounds, g_low, g_high = fit_group_bounds(
        df_train[~df_train['property_type_name'].isin(room_zero_groups)],
        col=col, group_col='property_type_name', q_low=0.005, q_high=0.995, min_count=100
    )
    room_bounds_dict[col] = (bounds, g_low, g_high)


def apply_room_clip(df_target):
    df_res = df_target.copy()
    
    land_mask = df_res['property_type_name'].isin(room_zero_groups)
    non_land = ~land_mask
    
    # 1. Gán cứng nhóm Đất về bằng 0
    df_res.loc[land_mask, room_cols] = 0

    # 2. Xử lý clip từng biến cho các nhóm còn lại
    for col in room_cols:
        bounds, g_low, g_high = room_bounds_dict[col]
        
        # Lấy biên dưới toán học học từ train (Chặn dưới tối thiểu là 1)
        lower_math = df_res.loc[non_land, 'property_type_name'].map(bounds['low']).fillna(g_low)
        lower = np.maximum(lower_math, 1)
        
        # Thay vì ép lấy np.minimum, ta tin tưởng hoàn toàn vào hạn mức thực tế đã tinh chỉnh
        upper = df_res.loc[non_land, 'property_type_name'].map(room_hard_high[col]).fillna(8)

        # Tiến hành ép biên an toàn
        df_res.loc[non_land, col] = df_res.loc[non_land, col].clip(lower=lower, upper=upper)
        
    # Làm tròn và ép kiểu số nguyên
    for col in room_cols:
        df_res[col] = np.round(df_res[col].fillna(0)).astype(int)
        
    return df_res

# Lưu cấu trúc dữ liệu trước xử lý để vẽ biểu đồ so sánh
df_train_before_room = df_train.copy()

# Áp dụng sạch cho cả hai tập dữ liệu
df_train = apply_room_clip(df_train)
df_test = apply_room_clip(df_test)

print("Ngưỡng bedroom_count học từ train:")
display(room_bounds_dict['bedroom_count'][0])
print("Ngưỡng bathroom_count học từ train:")
display(room_bounds_dict['bathroom_count'][0])

# =====================================================================
# HIỂN THỊ SƠ ĐỒ SO SÁNH PHÂN NHÓM THEO TYPE (THAY THẾ HÀM CŨ)
# =====================================================================
import matplotlib.pyplot as plt
import seaborn as sns

for col in room_cols:
    fig, axes = plt.subplots(1, 2, figsize=(16, 5), sharey=True)
    
    # Biểu đồ trái: Trước xử lý phân nhóm
    sns.boxplot(data=df_train_before_room, x=col, y='property_type_name', ax=axes[0], palette='Set3')
    axes[0].set_title(f'{col.upper()} Trước Xử Lý (Phân Nhóm)', fontsize=12, fontweight='bold')
    axes[0].set_xlabel('Số lượng phòng (Gốc)')
    axes[0].set_ylabel('Loại hình bất động sản')
    
    # Biểu đồ phải: Sau xử lý phân nhóm
    sns.boxplot(data=df_train, x=col, y='property_type_name', ax=axes[1], palette='Set3')
    axes[1].set_title(f'{col.upper()} Sau Xử Lý (Phân Nhóm)', fontsize=12, fontweight='bold')
    axes[1].set_xlabel('Số lượng phòng (Gốc)')
    axes[1].set_ylabel('')
    
    plt.tight_layout()
    plt.show()
```

Ví dụ: `house_direction` có thể có hàng chục giá trị hiếm gặp; `ward_name` có hàng trăm phường chỉ xuất hiện vài lần.

**Chiến lược:** Gộp các category thưa về nhãn `'Khác'` để giảm nhiễu, học ngưỡng từ train rồi áp lên test.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# =========================================================
# HÀM GỘP LONG-TAIL CATEGORIES
# =========================================================

def fit_frequent_categories(train_df, col, min_freq=0.01):
    """
    Học danh sách category đủ phổ biến từ train.
    Các category chiếm < min_freq (tỉ lệ) được gộp về 'Khác'.
    min_freq=0.01 nghĩa là category phải chiếm ít nhất 1% tổng dữ liệu.
    """
    freq = train_df[col].value_counts(normalize=True)
    frequent = set(freq[freq >= min_freq].index.tolist())
    frequent.add('Unknown')  # Giữ lại Unknown đã tạo trước đó
    return frequent

def apply_rare_category_replace(df, col, frequent_set, rare_label='Khác'):
    """
    Áp dụng gộp category thưa. Chỉ thay những category KHÔNG nằm trong frequent_set.
    """
    df = df.copy()
    df[col] = df[col].apply(lambda x: x if x in frequent_set else rare_label)
    return df

# =========================================================
# PHÂN TÍCH LONG-TAIL THEO TỪNG BIẾN CATEGORICAL
# =========================================================

cat_cols_to_check = ['house_direction', 'ward_name', 'district_name', 'province_name']

print("=== PHÂN TÍCH LONG-TAIL CATEGORIES ===\n")

longtail_summary = []
for col in cat_cols_to_check:
    if col not in df_train.columns:
        continue
    vc = df_train[col].value_counts(normalize=True)
    n_total = len(vc)
    n_rare = (vc < 0.01).sum()
    pct_rare = round(n_rare / n_total * 100, 1)
    longtail_summary.append({
        'Biến': col,
        'Tổng số category': n_total,
        'Category hiếm (< 1%)': n_rare,
        'Tỉ lệ hiếm (%)': pct_rare
    })
    print(f"  {col}: {n_total} categories, {n_rare} hiếm ({pct_rare}%)")
    print(f"    Top 5: {list(vc.head().index)}")
    print(f"    Bottom 5: {list(vc.tail().index)}\n")

display(pd.DataFrame(longtail_summary))


# =========================================================
# VẼ PARETO CHART CHO house_direction (biến hay bị long-tail)
# =========================================================
if 'house_direction' in df_train.columns:
    vc_dir = df_train['house_direction'].value_counts(normalize=True)
    
    fig, ax = plt.subplots(figsize=(12, 4))
    vc_dir.plot(kind='bar', ax=ax, color='steelblue')
    ax.axhline(0.01, color='red', linestyle='--', label='Ngưỡng 1% (rare)')
    ax.set_title('Phân phối house_direction — Pareto', fontweight='bold')
    ax.set_ylabel('Tỉ lệ (%)')
    ax.legend()
    plt.tight_layout()
    plt.show()

# =========================================================
# ÁP DỤNG GỘP LONG-TAIL
# =========================================================
def clean_street(text):
    if pd.isna(text): return 'Unknown'
    return str(text).lower().strip().replace('đường ', '').replace('đg ', '')

df_train['street_name'] = df_train['street_name'].apply(clean_street)
df_test['street_name'] = df_test['street_name'].apply(clean_street)
# Chỉ xử lý những biến có long-tail rõ ràng, bỏ qua province/district vì cần cho phân tích địa lý
cols_to_group_rare = [c for c in ['house_direction', 'ward_name', "street_name"] if c in df_train.columns]

freq_sets = {}
for col in cols_to_group_rare:
    min_f = 0.005 if col == 'street_name' else 0.01
    freq_sets[col] = fit_frequent_categories(df_train, col, min_freq=min_f)
    n_before = df_train[col].nunique()
    df_train = apply_rare_category_replace(df_train, col, freq_sets[col])
    df_test  = apply_rare_category_replace(df_test,  col, freq_sets[col])
    n_after  = df_train[col].nunique()
    print(f"✅ {col}: {n_before} → {n_after} categories (gộp {n_before - n_after} nhóm thưa về 'Khác')")

print("\n=== KIỂM TRA SAU GỘP ===")
for col in cols_to_group_rare:
    print(f"{col} — top categories train:", df_train[col].value_counts().head(5).to_dict())
```

###  Phát hiện Outlier Đa Biến (Multivariate Outlier Detection)

Xử lý univariate (từng biến riêng lẻ) có thể bỏ sót các tổ hợp bất thường như:
- Diện tích 200 m² nhưng giá chỉ 200 triệu → `price_per_sqm` cực thấp
- 10 phòng ngủ nhưng diện tích 25 m² → phi lý về mặt vật lý
- Nhà 1 m² (lỗi nhập liệu) có giá bình thường → thoát qua univariate nhưng vẫn lỗi

**Phương pháp:** Kết hợp kiểm tra tỉ số đơn giản (sanity ratio) và Isolation Forest để phát hiện outlier đa chiều.
Trong bài toán phân cụm, không nên xóa các điểm này ngay; nên **gắn cờ** để theo dõi.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

# =========================================================
# BƯỚC 1: SANITY RATIO CHECKS — Kiểm tra logic thực tế
# =========================================================
print("=== BƯỚC 1: SANITY RATIO CHECKS ===\n")

df_train_mv = df_train.copy()

# Check 1: price_per_sqm bất thường (< 1 triệu/m² hoặc > 1 tỷ/m²)
ppsqm_too_low  = df_train_mv['price_per_sqm'] < 1_000_000
ppsqm_too_high = df_train_mv['price_per_sqm'] > 1_000_000_000
print(f"price_per_sqm < 1 triệu/m²   : {ppsqm_too_low.sum():,} dòng")
print(f"price_per_sqm > 1 tỷ/m²      : {ppsqm_too_high.sum():,} dòng")

# Check 2: Phòng ngủ/diện tích phi lý (> 1 phòng ngủ / 10m²)
if 'bedroom_count' in df_train_mv.columns:
    room_density_flag = (df_train_mv['bedroom_count'] > 0) &                         (df_train_mv['area'] / df_train_mv['bedroom_count'] < 10)
    print(f"Mật độ phòng > 1 phòng/10m²  : {room_density_flag.sum():,} dòng")

# Check 3: Đất có phòng ngủ > 0 (đã xử lý ở trên nhưng check lại)
if 'property_type_name' in df_train_mv.columns:
    land_with_rooms = (df_train_mv['property_type_name'] == 'Đất') &                       (df_train_mv['bedroom_count'] > 0)
    print(f"Đất có phòng ngủ > 0          : {land_with_rooms.sum():,} dòng")

# =========================================================
# BƯỚC 2: ISOLATION FOREST — Phát hiện outlier đa chiều
# =========================================================
print("\n=== BƯỚC 2: ISOLATION FOREST ===\n")

# Chọn các feature số để detect outlier đa biến
iso_features = ['price', 'area', 'price_per_sqm']
iso_features = [f for f in iso_features if f in df_train_mv.columns]

# Lấy dữ liệu không có null
iso_data = df_train_mv[iso_features].dropna()
iso_idx = iso_data.index

# Scale về cùng thang đo (log1p để giảm lệch)
iso_log = np.log1p(iso_data)
scaler = StandardScaler()
iso_scaled = scaler.fit_transform(iso_log)

# Fit Isolation Forest (contamination=0.02 nghĩa là kỳ vọng ~2% outlier)
iso_forest = IsolationForest(
    n_estimators=200,
    contamination=0.02,
    random_state=42,
    n_jobs=-1
)
iso_pred = iso_forest.fit_predict(iso_scaled)
iso_score = iso_forest.decision_function(iso_scaled)

# Gắn kết quả về df_train
df_train_mv['iso_flag'] = 0
df_train_mv.loc[iso_idx, 'iso_flag'] = (iso_pred == -1).astype(int)
df_train_mv.loc[iso_idx, 'iso_score'] = iso_score

n_outliers = (df_train_mv['iso_flag'] == 1).sum()
print(f"Isolation Forest phát hiện: {n_outliers:,} điểm bất thường ({n_outliers/len(df_train_mv)*100:.2f}%)")

# =========================================================
# BƯỚC 3: VISUALIZATION — Outlier đa biến
# =========================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

sample_mv = df_train_mv.sample(min(30000, len(df_train_mv)), random_state=42)

# Plot 1: price vs area — tô màu outlier
normal = sample_mv[sample_mv['iso_flag'] == 0]
outlier = sample_mv[sample_mv['iso_flag'] == 1]

axes[0].scatter(np.log1p(normal['area']), np.log1p(normal['price']),
                c='steelblue', alpha=0.3, s=5, label='Bình thường')
axes[0].scatter(np.log1p(outlier['area']), np.log1p(outlier['price']),
                c='red', alpha=0.7, s=20, label='Outlier đa biến')
axes[0].set_xlabel('log1p(area)')
axes[0].set_ylabel('log1p(price)')
axes[0].set_title('Phát hiện Outlier đa biến: Price vs Area', fontweight='bold')
axes[0].legend()

# Plot 2: Phân bố outlier theo loại BĐS
outlier_by_type = df_train_mv.groupby('property_type_name')['iso_flag'].mean() * 100
outlier_by_type.sort_values(ascending=False).plot(kind='bar', ax=axes[1], color='salmon')
axes[1].set_title('Tỉ lệ outlier đa biến theo loại BĐS (%)', fontweight='bold')
axes[1].set_ylabel('Tỉ lệ (%)')
axes[1].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()

# =========================================================
# BƯỚC 4: QUYẾT ĐỊNH XỬ LÝ — GẮN CỜ, KHÔNG XÓA
# =========================================================
# Trong bài toán clustering, KHÔNG nên xóa outlier đa biến vì chúng
# có thể là một cụm riêng biệt (luxury, đất đặc biệt...).
# Gắn flag để theo dõi và có thể phân tích riêng sau khi clustering.

df_train['iso_outlier_flag'] = df_train_mv['iso_flag'].values
df_train['iso_score'] = df_train_mv['iso_score'].values

# Áp scaler học từ train lên test
iso_data_test = df_test[iso_features].fillna(df_test[iso_features].median())
iso_log_test = np.log1p(iso_data_test)
iso_scaled_test = scaler.transform(iso_log_test)
df_test['iso_outlier_flag'] = (iso_forest.predict(iso_scaled_test) == -1).astype(int)
df_test['iso_score'] = iso_forest.decision_function(iso_scaled_test)

print("\n✅ Đã gắn cờ outlier đa biến vào cả train và test.")
print(f"Train - Outlier flag = 1: {df_train['iso_outlier_flag'].sum():,} dòng")
print(f"Test  - Outlier flag = 1: {df_test['iso_outlier_flag'].sum():,} dòng")
print("\n⚠️  Lưu ý: Các điểm bị gắn cờ KHÔNG bị xóa. Sau khi clustering xong,")
print("   hãy phân tích xem chúng thuộc cụm nào và có ý nghĩa kinh tế không.")
```

### Phân tích Tác động của Outlier lên Clustering (Impact Analysis)

Bước này trả lời câu hỏi quan trọng: **"Việc xử lý outlier thực sự cải thiện chất lượng clustering không?"**

Phương pháp: Chạy K-Means nhanh trên dữ liệu trước và sau xử lý outlier, so sánh:
- Inertia (tổng khoảng cách nội cụm)
- Silhouette Score (mức độ phân tách cụm)
- Độ ổn định centroid (variance)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans, MiniBatchKMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

# =========================================================
# CHUẨN BỊ FEATURES ĐƠN GIẢN CHO IMPACT TEST
# =========================================================
# Dùng một subset feature số để so sánh nhanh
impact_features = ['price', 'area', 'price_per_sqm']
impact_features = [f for f in impact_features if f in df_train.columns]

K_TEST = 5  # Số cụm thử nghiệm

def prepare_for_clustering(df, features, log_transform=True, sample_n=30000):
    """Chuẩn bị data sạch, log-transform và scale cho clustering thử nghiệm."""
    data = df[features].dropna().copy()
    data = data.sample(min(sample_n, len(data)), random_state=42)
    if log_transform:
        data = np.log1p(data)
    scaler = StandardScaler()
    return scaler.fit_transform(data), data.index

def run_kmeans_quick(data_scaled, k, random_state=42):
    """Chạy MiniBatchKMeans nhanh và trả về inertia + silhouette."""
    km = MiniBatchKMeans(n_clusters=k, random_state=random_state, n_init=3, batch_size=2048)
    labels = km.fit_predict(data_scaled)
    inertia = km.inertia_
    # Silhouette trên sample nhỏ để tăng tốc
    sample_idx = np.random.choice(len(data_scaled), min(5000, len(data_scaled)), replace=False)
    sil = silhouette_score(data_scaled[sample_idx], labels[sample_idx])
    return inertia, sil, labels

print("=== IMPACT ANALYSIS: OUTLIER vs CLUSTERING ===\n")

# =========================================================
# TẬP DỮ LIỆU TRƯỚC XỬ LÝ (dùng backup đã lưu)
# =========================================================
# Thêm price_per_sqm vào backup nếu chưa có
if 'price_per_sqm' not in df_train_before_outlier.columns:
    df_before = df_train_before_outlier.copy()
    df_before['price_per_sqm'] = df_before['price'] / df_before['area']
else:
    df_before = df_train_before_outlier.copy()

# =========================================================
# CHẠY KMEANS TRƯỚC/SAU
# =========================================================
data_before, _ = prepare_for_clustering(df_before, impact_features)
data_after,  _ = prepare_for_clustering(df_train,  impact_features)

inertia_before, sil_before, labels_before = run_kmeans_quick(data_before, K_TEST)
inertia_after,  sil_after,  labels_after  = run_kmeans_quick(data_after,  K_TEST)

print(f"{'Chỉ số':<30} {'Trước xử lý':>15} {'Sau xử lý':>15} {'Cải thiện':>12}")
print("-" * 75)
print(f"{'Inertia':<30} {inertia_before:>15,.0f} {inertia_after:>15,.0f} "
      f"{(inertia_before-inertia_after)/inertia_before*100:>11.1f}%")
print(f"{'Silhouette Score':<30} {sil_before:>15.4f} {sil_after:>15.4f} "
      f"{(sil_after-sil_before)/abs(sil_before)*100:>11.1f}%")

# =========================================================
# VISUALIZATION: So sánh phân tách cụm
# =========================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

for ax, data, labels, title in zip(
    axes,
    [data_before, data_after],
    [labels_before, labels_after],
    ['Phân cụm TRƯỚC xử lý outlier', 'Phân cụm SAU xử lý outlier']
):
    scatter = ax.scatter(data[:, 0], data[:, 1], c=labels, cmap='tab10',
                         alpha=0.4, s=5)
    ax.set_title(title, fontweight='bold')
    ax.set_xlabel('log1p(price) [scaled]')
    ax.set_ylabel('log1p(area) [scaled]')
    plt.colorbar(scatter, ax=ax, label='Cluster')

plt.suptitle(f'K-Means K={K_TEST}: Phân cụm Trước vs Sau xử lý Outlier',
             fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()

# =========================================================
# ELBOW CHART — Chọn K tối ưu trên dữ liệu đã xử lý
# =========================================================
print("\n⏳ Đang chạy Elbow chart để chọn K tối ưu...")

k_range = range(2, 11)
inertias = []
silhouettes = []

for k in k_range:
    ine, sil, _ = run_kmeans_quick(data_after, k)
    inertias.append(ine)
    silhouettes.append(sil)

fig, axes = plt.subplots(1, 2, figsize=(14, 4))

axes[0].plot(k_range, inertias, marker='o', color='steelblue')
axes[0].set_title('Elbow Method — Inertia', fontweight='bold')
axes[0].set_xlabel('Số cụm K')
axes[0].set_ylabel('Inertia')

axes[1].plot(k_range, silhouettes, marker='s', color='coral')
axes[1].set_title('Silhouette Score theo K', fontweight='bold')
axes[1].set_xlabel('Số cụm K')
axes[1].set_ylabel('Silhouette Score')

plt.suptitle('Chọn K tối ưu trên dữ liệu sau xử lý outlier', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()

best_k = k_range.start + silhouettes.index(max(silhouettes))
print(f"\n✅ K tối ưu theo Silhouette Score: K = {best_k}")
print(f"   Silhouette Score tốt nhất: {max(silhouettes):.4f}")
```

Kết quả xử lý outlier có tác động tích cực rõ ràng lên chất lượng phân cụm: Inertia giảm 11% và Silhouette Score tăng 4.7% so với dữ liệu chưa xử lý. Các cụm sau xử lý phân tách rõ ràng và có ý nghĩa hơn, không còn bị kéo lệch bởi các giá trị cực đoan.

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

print("=== BƯỚC 5: CHUẨN HÓA CÁC BIẾN SỐ (NUMERICAL FEATURES) ===\n")

# Phân loại các cột số học dựa trên đặc thù phân phối
# 1. Nhóm lệch phải mạnh (Cần Log-transform + StandardScaler)
skewed_cols = ['price', 'area', 'price_per_sqm', 'frontage_width']

# 2. Nhóm rời rạc/giới hạn trần cứng (Linear) -> CHỈ CẦN StandardScaler
linear_cols = ['bedroom_count', 'bathroom_count', 'floor_count']

# Lọc lại danh sách cột để đảm bảo không bị lỗi KeyError nếu thiếu cột
skewed_cols = [col for col in skewed_cols if col in df_train.columns]
linear_cols = [col for col in linear_cols if col in df_train.columns]

# ==========================================
# PHẦN 1: XỬ LÝ NHÓM LỆCH PHẢI (SKEWED)
# ==========================================
print(f"Tiến hành Log Transform & Scale cho: {skewed_cols}")

# Khởi tạo Scaler riêng cho nhóm Log
scaler_skewed = StandardScaler()

# Tạo danh sách tên cột mới chứa hậu tố '_log_scaled' để không ghi đè dữ liệu gốc ngay lập tức (dễ debug)
skewed_scaled_cols = [f"{col}_log_scaled" for col in skewed_cols]

# Bước 1.1: Log Transform (dùng np.log1p để xử lý an toàn giá trị 0 nếu có)
df_train_log = np.log1p(df_train[skewed_cols])
df_test_log = np.log1p(df_test[skewed_cols])

# Bước 1.2: Chuẩn hóa Z-score (Fit trên Train, Transform trên Train & Test)
df_train[skewed_scaled_cols] = scaler_skewed.fit_transform(df_train_log)
df_test[skewed_scaled_cols] = scaler_skewed.transform(df_test_log)


# ==========================================
# PHẦN 2: XỬ LÝ NHÓM TUYẾN TÍNH (LINEAR)
# ==========================================
print(f"Tiến hành Scale trực tiếp cho: {linear_cols}")

# Khởi tạo Scaler riêng biệt
scaler_linear = StandardScaler()

# Tạo danh sách tên cột mới
linear_scaled_cols = [f"{col}_scaled" for col in linear_cols]

# Chuẩn hóa Z-score trực tiếp (Fit trên Train, Transform trên Train & Test)
if linear_cols:
    df_train[linear_scaled_cols] = scaler_linear.fit_transform(df_train[linear_cols])
    df_test[linear_scaled_cols] = scaler_linear.transform(df_test[linear_cols])

print("\n✅ Chuẩn hóa biến số hoàn tất!")
print("Các cột đã sẵn sàng cho mô hình FAMD + KMeans:")
print(" - Nhóm Skewed:", skewed_scaled_cols)
print(" - Nhóm Linear:", linear_scaled_cols)

# (Tùy chọn) Kiểm tra lại phương sai và trung bình sau khi chuẩn hóa trên tập Train
# Kỳ vọng: Trung bình ~ 0, Độ lệch chuẩn ~ 1
# print("\nKiểm tra thống kê trên tập Train (kỳ vọng mean~0, std~1):")
# display(df_train[skewed_scaled_cols + linear_scaled_cols].describe().round(3).loc[['mean', 'std']])
```

### 2.Các biến danh mục

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Giả định: Bạn đã có sẵn cột 'price_per_sqm' trong dữ liệu. 
# Nếu chưa có, bạn có thể tính bằng: df_train['price'] / df_train['area']

# ==========================================
# 1. NHÓM LOW CARDINALITY (Dành cho luồng MCA của FAMD)
# ==========================================
# Ép kiểu dữ liệu về chuỗi (string) để thư viện FAMD nhận diện là biến phân loại
categorical_mca_cols = ['ward_name', 'house_direction','ward_name', 'street_name']

for col in categorical_mca_cols:
    df_train[col] = df_train[col].astype(str)
    df_test[col] = df_test[col].astype(str)


# ==========================================
# 2. PROVINCE_NAME: FREQUENCY ENCODING
# ==========================================
# Tính tỷ lệ phần trăm (tần suất) xuất hiện của từng tỉnh trên df_train
province_freq = df_train['province_name'].value_counts(normalize=True)

# Map giá trị vào df_train và df_test
df_train['province_encoded'] = df_train['province_name'].map(province_freq)

# Nếu df_test có tỉnh lạ chưa từng xuất hiện trong tập train, gán tần suất = 0
df_test['province_encoded'] = df_test['province_name'].map(province_freq).fillna(0)


# ==========================================
# 3. DISTRICT_NAME: TARGET ENCODING
# ==========================================
# Tính trung vị (median) của giá trên mỗi m2 cho từng quận/huyện trên df_train
district_target_median = df_train.groupby('district_name')['price_per_sqm'].median()

# Tính trung vị của toàn bộ tập train để dự phòng
overall_median = df_train['price_per_sqm'].median()

# Map giá trị vào df_train và df_test
df_train['district_encoded'] = df_train['district_name'].map(district_target_median)

# Nếu df_test có quận/huyện lạ, gán bằng trung vị của toàn thị trường (overall_median)
df_test['district_encoded'] = df_test['district_name'].map(district_target_median).fillna(overall_median)


# ==========================================
# 4. STANDARD SCALER CHO BIẾN VỪA MÃ HÓA
# ==========================================
# Đưa các biến đã chuyển thành số về cùng thang đo (Mean=0, Std=1) để đưa vào luồng PCA
scaler = StandardScaler()
encoded_cols = ['province_encoded', 'district_encoded']

# FIT trên df_train, sau đó TRANSFORM trên cả 2 tập
df_train[encoded_cols] = scaler.fit_transform(df_train[encoded_cols])
df_test[encoded_cols] = scaler.transform(df_test[encoded_cols])
```

```python
# Chọn ra danh sách các cột đã được chuẩn hóa để gửi cho người chạy mô hình
final_features = skewed_scaled_cols + linear_scaled_cols + encoded_cols + categorical_mca_cols

print("="*60)
print("📦 THÔNG TIN BỘ DỮ LIỆU ĐÃ CHUẨN HÓA (ĐẦU VÀO CHO FAMD)")
print("="*60)

# 1. In ra danh sách các phân nhóm biến
print("\n1. LUỒNG PCA (Các biến liên tục đã Scale, Mean~0, Std=1):")
print(" - Nhóm Skewed (Log + Scale) :", skewed_scaled_cols)
print(" - Nhóm Linear (Chỉ Scale)   :", linear_scaled_cols)
print(" - Nhóm Encoded (Mã hóa vị trí):", encoded_cols)

print("\n2. LUỒNG MCA (Các biến Categorical nguyên bản, Dạng String):")
print(" - Nhóm phân loại            :", categorical_mca_cols)

# 2. In thống kê mô tả (Mean, Std, Min, Max) để đối soát
print("\n" + "="*60)
print("📊 BẢNG THỐNG KÊ CÁC BIẾN SỐ HỌC (LUỒNG PCA)")
print("="*60)
numeric_ready_cols = skewed_scaled_cols + linear_scaled_cols + encoded_cols
display_stats = df_train[numeric_ready_cols].describe().loc[['mean', 'std', 'min', 'max']].round(3)
display(display_stats)

# 3. In ra 5 dòng đầu tiên xem thực tế định dạng dữ liệu
print("\n" + "="*60)
print("👀 PREVIEW 5 DÒNG DỮ LIỆU ĐẦU TIÊN")
print("="*60)
display(df_train[final_features].head())
```

```python
import pandas as pd
import re

print("=== BƯỚC 6: LÀM SẠCH VÀ BÓC TÁCH TỪ KHÓA TỪ METADATA ===\n")
df_train_copy_vector = df_train.copy()

# BƯỚC 1.1: ĐỊNH NGHĨA HÀM LÀM SẠCH VĂN BẢN
def clean_text_bds(text):
    if pd.isna(text):
        return ""
    
    # 1. Đưa về chữ thường
    text = str(text).lower()
    
    # 2. Xóa ký tự đặc biệt, dấu câu, emoji (chỉ giữ lại chữ cái, số và khoảng trắng)
    # Lưu ý: \w trong Python tự động giữ lại cả các ký tự tiếng Việt có dấu.
    text = re.sub(r'[^\w\s]', ' ', text)
    
    # 3. Xóa các khoảng trắng thừa (ví dụ: biến 3 dấu cách thành 1)
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

# BƯỚC 1.2: GỘP VÀ ÁP DỤNG LÀM SẠCH
print("Đang dọn dẹp ký tự đặc biệt trên hàng triệu dòng...")
# Nối chuỗi trước
df_train_copy_vector['full_text'] = df_train_copy_vector['name'].fillna('') + ' ' + df_train['description'].fillna('')
#df_test['full_text'] = df_test['name'].fillna('') + ' ' + df_test['description'].fillna('')

# Áp dụng hàm làm sạchdf_train_copy_vector
df_train_copy_vector['full_text'] = df_train_copy_vector['full_text'].apply(clean_text_bds)
#df_test['full_text'] = df_test['full_text'].apply(clean_text_bds)


# BƯỚC 2: KHAI BÁO TỪ ĐIỂN REGEX "ĂN TIỀN" (Không thay đổi)
regex_dict = {
    'is_phap_ly_chuan': r'(sổ hồng|sổ đỏ|pháp lý chuẩn|bao sang tên|hoàn công|đã có sổ)',
    'is_chinh_chu': r'(chính chủ|1 đời chủ|một đời chủ)',
    'is_hem_xe_hoi': r'(hẻm xe hơi|hxh|ô tô|xe hơi|đường \d+m|hẻm \d+m)',
    'is_mat_tien': r'(mặt tiền|mặt phố)',
    'is_lo_goc': r'(lô góc|2 mặt tiền|hai mặt tiền)',
    'is_kinh_doanh': r'(kinh doanh|buôn bán|sầm uất|mở văn phòng|spa)',
    'is_dong_tien': r'(dòng tiền|cho thuê|chdv|căn hộ dịch vụ|phòng trọ|thu nhập)',
    'is_nha_moi': r'(nhà mới|xách vali|ở ngay|full nội thất|tặng nội thất|cao cấp)',
    'is_tien_xay_moi': r'(nhà nát|tiện xây mới|đất trống|giấy phép xây dựng|đập đi)'
}


# BƯỚC 3: QUÉT VÀ TẠO CỘT MỚI
print("Đang bóc tách từ khóa bằng Regex...")
new_text_flags = list(regex_dict.keys())

for col_name, pattern in regex_dict.items():
    # Quét str.contains, ép kiểu map True -> 'Yes', False -> 'No'
    df_train_copy_vector[col_name] = df_train_copy_vector['full_text'].str.contains(pattern, na=False).map({True: 'Yes', False: 'No'})
    #df_test[col_name] = df_test['full_text'].str.contains(pattern, na=False).map({True: 'Yes', False: 'No'})

# Dọn dẹp RAM: Xóa cột trung gian và các cột văn bản gốc
#df_train = df_train.drop(columns=['full_text', 'name', 'description'], errors='ignore')
#df_test = df_test.drop(columns=['full_text', 'name', 'description'], errors='ignore')

print(f"✅ Đã bóc tách thành công {len(new_text_flags)} cờ (flags) từ văn bản sạch!")
```
